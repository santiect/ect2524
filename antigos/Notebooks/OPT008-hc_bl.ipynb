````python
import tsplib95
import numpy as np
import sys
import wget
import os

class DataLoader:

    instances = {
        'dj38'  : 'https://www.math.uwaterloo.ca/tsp/world/dj38.tsp',
        'lux980': 'https://www.math.uwaterloo.ca/tsp/world/lu980.tsp',
        'mu1979': 'https://www.math.uwaterloo.ca/tsp/world/mu1979.tsp',
        'uy734' : 'https://www.math.uwaterloo.ca/tsp/world/uy734.tsp',
        'ar9152': 'https://www.math.uwaterloo.ca/tsp/world/ar9152.tsp',
        'ja9847': 'https://www.math.uwaterloo.ca/tsp/world/ja9847.tsp',
    }

    def get_matrix(id):
        try:
            tmp_filename = '.tmp.tsp'
            wget.download(DataLoader.instances[id], tmp_filename)
            problem = tsplib95.load('.tmp.tsp')
            coords = problem.node_coords
            coord = np.array(list(coords.values()))
            x = np.repeat(coord, coord.shape[0], axis=1).T.reshape((coord.shape[1], coord.shape[0], coord.shape[0]))
            x = (x-np.transpose(x, (0, 2, 1)))**2
            dist = np.trunc(np.round(np.sqrt(np.sum(x, axis=0))))
            os.remove(tmp_filename)
            return dist
        except:
            print(f"Unable to load {id}")
            return None

matrix = DataLoader.get_matrix('ja9847')
matrix
````
