# O que significa otimização?

Um problema de otimização pode ser expresso como:

$$
    \text{min} = f(x) \tag{1}
$$

sujeito a

$$
    x \in S \tag{2}
$$

em que:
* $x$ é a **solução** do problema;
* $f(x)$ é a **função objetivo**, e;
* $S$ é o **conjunto** de soluções **viáveis** do problema;

A **função objetivo** é uma função matemática, ou expressão, que representa uma ou mais grandezas, cujo valor mínimo (ou máximo) queremos encontrar determinando os valores de $x$. Neste contexto, $x$ representa o conjunto de **variáveis de decisão** ou **incógnitas**. A função objetivo pode representar, por exemplo:

* Distância para se deslocar de uma cidade até a outra;
* Desperdício de material para a fabricação de um item;
* Gasto com a instalação de cabos de dados submarinos;
* Tempo total de execução de tarefas;
* Tempo total de uma equipe de natação em uma prova de revezamento;
* Dias trabalhados;
* Quantidade de horários de aula cobertos por um conjunto de professores;
* Quantidade de cores em uma imagem;
* Similaridade entre perfis de consumidores;
* Custo de construção de uma rede de distribuição de água;
* Tempo de espera por uma ambulância;
* Eficiência de um sistema físico;
* Entre outros...

O conjunto de **soluções viáveis** pode ser delimitado por um **conjunto de funções**, que definem quais soluções são válidas (**viáveis**) para um problema. Este conjunto também é chamado de **conjunto de restrições**. Existem problemas que possuem **infinitas soluções viáveis**, assim como existem problemas com um **conjunto finito soluções**.

Quando $S=\emptyset$, dizemos que o problema é de otimização **irrestrita**. Porém, se $S\neq \emptyset$, então o problema é de otimização **restrita**.

*Nota*: um problema restrito pode ter infinitas soluções.



#Modelagem

Qualquer situação do mundo real que possa ser representada por expressões matemáticas pode ser tratada com otimização, desde que o modelo formulado permita uma aproximação satisfatória da realidade (problema) face ao propósito de aplicação.

Logo, um **modelo** é uma representação do mundo real e ao considerar otimização estamos falando em um **modelo simbólico**, expresso na forma de parâmetros, variáveis, índices, funções, igualdades, desigualdes e etc.

**Modelar** é um misto de **arte e ciência**, pois exige criatividade para observar, entender e traduzir a realidade. Porém, sem algum **conhecimento técnico e científico**, estaríamos impossibilitados de tratar adequadamente um problema real para o qual buscamos solução.

#Ferramentas computacionais

As duas principais ferramentas que podemos ter à mão para trabalhar com otimização de forma aplicada, ou mesmo prototipar uma ideia rapidamente são:

* **Linguagem de modelagem**: permite escrever a formulação de nosso modelo em um programa de computador de forma similar a como o criamos usando apenas notação matemática;

* **Resolvedor ou *solver***: programa especializado que, dados os parâmetros do problema (dados de entrada), bem como sua formulação, possui algoritmos específicos que buscam encontrar as soluções (valores das variáveis de decisão).

Dentre algumas das linguagens de modelagem, podemos citar AMPL, GAMS, ZIMPL e OPL. Alguns resolvedores populares são CPLEX, Gurobi, Couenne, Bonmin e snopt. O próprio Excell permite resolver problemas de otimização. A linguagem de modelagem nos permite ao mesmo tempo especificar um modelo matemático e interagir com um ou mais resolvedores para buscar as soluções.

Neste curso usaremos AMPL, uma linguagem simples de aprender, com versão gratuita, e que permite utilizar diversos resolvedores disponíveis no mercado.

> ***Nota***: a maioria, senão todas as linguagens de programação, possuem bibliotecas para otimização, ou mesmo APIs disponibilizadas pelos criadores dos resolvedores. Desta forma, qualquer software que é produzido pode se utilizar destes recursos, seja em Python, C/C++, Java, Javascript, PHP, etc.

> ***Nota***: AMPL possui uma API para linguagens de programação de propósito geral. Podemos por exemplo criar um modelo em AMPL e executá-lo a partir do Python. A vantagem em utilizar esta abordagem é que podemos escrever os modelos sempre usando a **mesma sintaxe** (AMPL), resolver o modelo com **solvers diferentes**, porém construindo um software em Python. Muitas vezes, a sintaxe muda muito, assim como o nível de complexidade para se utilizar APIs fornecidas por diferentes vendedores de resolvedores.

> ***Nota:*** Mesmo que desável, não é necessário ser programador para construir e aplicar modelos de otimização. Pode-se resolver muitas situações do dia-a-dia de uma empresa apenas com conhecimentos em modelagem.


#Um "HelloWorld" diferente

Suponha que você irá acampar no final de semana com alguns amigos. Procurando por seus materias de acampamento, acabou encontrando uma mochila com capacidade de carga de 10Kg.

Durante o processo de busca, encontrou também 6 itens que poderá levar nesta mochila. Uma consideração importante sobre estes itens é que ao somar o peso de cada um, o peso total ultrapassou a capacidade de carga da mochila. Logo, não será possível levar todos estes itens.

Para lhe auxiliar no processo de decisão de quais itens deverá levar, então, estabeleceu um sistema da pontuação, com notas de 1 a 10. Neste sistema, 10 significa que o item é crucial para o sucesso do acampamento, enquanto que 1 representa que o acampamento ocorreria sem maiores transtornos caso o item não seja levado. A lista dos itens disponíveis é a seguinte:

* Item 1: Lanterna (1Kg, 8 pontos)
* Item 2: Barraca (4Kg, 10 pontos)
* Item 3: Reservatório de água (3Kg, 10 pontos)
* Item 4: Reservatório de água (2Kg, 8 pontos)
* Item 5: Colchão Inflável (0.5Kg, 7 pontos)
* Item 6: JBL (2Kg, 4 pontos)

A solução do problema descrito consiste em decidir (**sim** ou **não**), se um item deverá ser levado na mochila considerando que queremos **maximizar** a soma das **pontuações** dos **itens escolhidos**. Ao mesmo tempo, a **combinação** de itens escolhidos **não** pode **ultrapassar** a **capacidade da mochila**.

Do parágrafo anterior, podemos identificar 3 elementos cruciais para a modelagem do nosso problema:

1. Podemos representar a decisão de levar ou não um determinado item por meio de uma **variável de decisão binária**, assumindo que esta variável assuma valor 1 quando a solução do problema indicar que o referido item deve ser colocado na mochila, e zero caso contrário;
2. A **função objetivo** a ser maximizada pode ser formulada como um **somatório**, considerando as variáveis de decisão  e a pontuação de cada item.
3. A **única restrição** para este problema pode ser modelada por uma desigualde, na qual o **somatório** dos pesos dos itens colocados na mochila não pode ser maior do que a capacidade desta.

> ***Nota***: os dados/valores que temos em mãos acerca do problema, chamamos de **Instância**. Estes dados, quando escritos na formulação matemática, chamamos de **parâmetros**.

##Usando AMPL para encontrar a solução do problema da mochila

Para utilizar a linguagem de modelagem, podemos seguir dois passos simples:

1.  Criar um arquivo de código fonte com a extensão .mod, dentro do qual escreveremos as insturções necessárias para descrever a formulação;
2. Criar um arquivo de código-fonte com a extensão .run, dentro do qual poderemos inserir as instruções para a resolução do modelo do arquivo .mod.

>**Nota**: a segunda etapa não é obrigatória, pois o AMPL possui um terminal interativo que pode ser utilizado no console do sistema. Neste terminal, podemos digitar os comandos para execução, um a um. A vantagem de ser criar o arquivo .run é que podemos automatizar a realização dos experimentos, não havendo necessidade de digitar repetidamente os mesmos comandos.


```python
%pip install -q amplpy
from amplpy import AMPL, ampl_notebook
ampl = ampl_notebook(
    modules=["highs", "cbc", "gurobi", "cplex"], # pick from over 20 modules including most commercial and open-source solvers
    license_uuid="499a8852-4e10-4d0b-a0b7-e45ad6f0c1b2") # your license UUID (e.g., free ampl.com/ce or ampl.com/courses licenses)
```

```python
%%writefile mochila.mod
var x1 binary;
var x2 binary;
var x3 binary;
var x4 binary;
var x5 binary;
var x6 binary;

maximize obj: 8*x1 + 10*x2 + 10*x3 + 8*x4 + 7*x5 + 100*x6;

subject to r1: x1 + 4*x2 + 3*x3 + 2*x4 + 0.5*x5 + 2*x6 <= 10;

#subject to r2: x2 == 1;

subject to r3: x1 + x2 + x3 + x4 + x6 == 5;

#subject to r2: 0 <= x1 <= 1;
#subject to r3: 0 <= x2 <= 1;
#subject to r4: 0 <= x3 <= 1;
#subject to r5: 0 <= x4 <= 1;
#subject to r6: 0 <= x5 <= 1;
#subject to r7: 0 <= x6 <= 1;
```

```python
%%writefile mochila.run
model "mochila.mod";
option solver highs;
solve;
```

```python
%%ampl_eval
reset;
include "mochila.run";
display x1, x2, x3, x4, x5, x6;
display obj;
```
